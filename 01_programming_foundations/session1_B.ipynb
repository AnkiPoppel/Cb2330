Harmonic Mean formula - used to compute an average for rates

$$
\text{Harmonic mean} =
\frac{N}{\displaystyle\sum_{i=1}^{N}\frac{1}{x_i}}
$$

```python
counts = [
    3473, 2876, 2480, 2445, 1524, 1980, 1237, 3635, 2060, 2386,
    2562, 2109, 1915, 2189, 3094, 1974, 2407, 2653, 2374, 1940,
    1871, 2456, 2140, 2207, 4757, 2882, 3834, 2512, 1663, 2023,
    1222, 591, 2266, 985, 2432, 3248, 2132, 1292, 2490, 3251,
    1979, 2000, 1736, 2612, 2411, 2283, 1324, 1948, 2084, 3840,
    2375, 2223, 2069, 973, 2088, 2541, 6317, 1918, 3863, 1647,
]
```

```python
def divide_total_reads(counts):
    total = 0
    for value in counts:
      total = total + 1/(value)
    return total
```

```python
def harmonic_mean_reads(counts):
    return len(counts) / divide_total_reads(counts)
```

```python
loop_res = len(counts) / divide_total_reads(counts)
```

```python
print(loop_res)
```

```python
import numpy as np
from scipy import stats
```

```python
numpy_res = (stats.hmean(counts))
print(numpy_res)
```

```python
np.isclose(numpy_res, loop_res)
```
